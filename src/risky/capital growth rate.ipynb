```python
import numpy as np
import pandas as pd

r = 0.1            
k = 50
gamma_values = [0.1, 0.2, 0.4, 0.8]  
n_trials = 40        

#np.random.seed(42)

g_results = np.zeros((n_trials, len(gamma_values)))
S_results = np.zeros((n_trials, len(gamma_values)))

for trial in range(n_trials):
    for idx, gamma in enumerate(gamma_values):
        #вектор
        z = np.random.rand(k)

        g_i = r + 2 * gamma * (z - 0.5)
        
        # g(k) = [(1+g1)(1+g2)...(1+gk)]^(1/k) - 1
        geometric_mean = np.prod(1 + g_i) ** (1.0 / k)
        g_k = geometric_mean - 1
    

        S_k = (1 + g_k) ** k
        
        g_results[trial, idx] = g_k
        S_results[trial, idx] = S_k

columns = [f'γ={g}' for g in gamma_values]
df_g = pd.DataFrame(g_results, columns=columns)
df_S = pd.DataFrame(S_results, columns=columns)

df_g.index += 1
df_S.index += 1

print("Таблица 1 – Темпы роста капитала g(k):")
print(df_g.to_string(float_format="{:.3f}".format))

print("\nТаблица 2 – Величина капитала S(k):")
print(df_S.to_string(float_format="{:.3f}".format))

df_g.to_csv('table1.csv')
df_S.to_csv('table2.csv')
```
